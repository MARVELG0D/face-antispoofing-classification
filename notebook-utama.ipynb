# **2. Offline Face Cropping dengan Auto-Rotation (YOLOv8-Face)**

Dalam eksperimen *Computer Vision* yang berfokus pada detail mikroskopis (seperti tekstur *Moiré* atau *blur* pada kasus *spoofing*), membiarkan model melihat seluruh area gambar (termasuk latar belakang yang tidak relevan) adalah sebuah kerugian besar. 

Latar belakang yang bising (*noisy background*) dapat menyebabkan model "menghafal" pola latar belakang kelas tertentu alih-alih mempelajari tekstur wajah yang sebenarnya.

Oleh karena itu, pada tahap ini kita mengimplementasikan **Data Purification Pipeline** menggunakan detektor wajah *State-of-the-Art*, yaitu **YOLOv8-Face**. 

Langkah ini dirancang dengan mekanisme pertahanan berlapis:
1. **Auto-Rotation (4 Skenario):** Banyak gambar dalam dataset yang tersimpan dengan orientasi terbalik atau menyamping (akibat orientasi EXIF kamera). Sistem ini secara otomatis memutar gambar ke 4 arah berbeda (0°, 90°, 180°, 270°) dan membiarkan YOLO memilih rotasi mana yang menghasilkan nilai *confidence* deteksi wajah tertinggi.
2. **Margin Cropping (30% Contextual Buffer):** Memotong wajah terlalu pas (*tight crop*) dapat menghilangkan informasi krusial di batas tepi wajah (seperti tepian topeng 3D atau tepian kertas foto). Kami menerapkan margin 30% untuk menyertakan sedikit konteks lokal.
3. **Offline Processing:** Alih-alih melakukan deteksi wajah secara *on-the-fly* di setiap iterasi *dataloader* (yang akan sangat memperlambat proses *training*), kami memproses seluruh gambar secara *offline* di awal, dan menyimpannya kembali ke *disk*.

Hasil akhirnya adalah sebuah dataset wajah yang bersih, terstandarisasi, menghadap ke arah yang benar, dan siap diolah oleh *Vision Transformer*.

```python
# Download YOLOv8-face Medium
!gdown --id 1IJZBcyMHGhzAi0G4aZLcqryqZSjPsps- -O yolov8m-face.pt

face_model = YOLO('yolov8m-face.pt')
print("YOLOv8-face loaded!")
```

```python
def process_and_crop_face(img_path, model, margin=CFG.margin, output_size=CFG.crop_size):
    img = cv2.imread(img_path)
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    rotations = [
        (img, 0),
        (cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE), 90),
        (cv2.rotate(img, cv2.ROTATE_180), 180),
        (cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE), 270)
    ]

    best_conf = -1.0
    best_box  = None
    best_img  = None

    for rot_img, angle in rotations:
        results = model(rot_img, verbose=False)[0]
        boxes   = results.boxes.data.cpu().numpy()
        if len(boxes) > 0:
            max_idx = np.argmax(boxes[:, 4])
            conf    = boxes[max_idx, 4]
            if conf > best_conf:
                best_conf = conf
                best_box  = boxes[max_idx, :4]
                best_img  = rot_img

    if best_box is not None:
        x1, y1, x2, y2 = best_box
        h_img, w_img   = best_img.shape[:2]
        w_box = x2 - x1
        h_box = y2 - y1
        x1 = max(0, int(x1 - margin * w_box))
        y1 = max(0, int(y1 - margin * h_box))
        x2 = min(w_img, int(x2 + margin * w_box))
        y2 = min(h_img, int(y2 + margin * h_box))
        crop_img = best_img[y1:y2, x1:x2]
    else:
        crop_img = img  # Fallback: gambar penuh

    final_img = cv2.resize(crop_img, (output_size, output_size))
    return final_img
```

```python
out_train_dir = "/kaggle/working/cropped_data/train"
out_test_dir  = "/kaggle/working/cropped_data/test"
os.makedirs(out_train_dir, exist_ok=True)
os.makedirs(out_test_dir,  exist_ok=True)

train_records = []

print("=== Memproses Data Train ===")
for cls_name in CFG.classes:
    cls_dir = os.path.join(CFG.train_dir, cls_name)
    if not os.path.exists(cls_dir):
        continue
    os.makedirs(os.path.join(out_train_dir, cls_name), exist_ok=True)

    for img_name in tqdm(os.listdir(cls_dir), desc=f"Cropping {cls_name}"):
        img_path     = os.path.join(cls_dir, img_name)
        cropped_img  = process_and_crop_face(img_path, face_model)
        if cropped_img is not None:
            save_path = os.path.join(out_train_dir, cls_name, img_name)
            cv2.imwrite(save_path, cv2.cvtColor(cropped_img, cv2.COLOR_RGB2BGR))
            train_records.append({
                'image_path': save_path,
                'label':      cls_name,
                'target':     CFG.class_to_idx[cls_name]
            })

train_df = pd.DataFrame(train_records)
train_df.to_csv("/kaggle/working/train_df.csv", index=False)
print(f"Total Train Images: {len(train_df)}")
print(train_df['label'].value_counts())

print("\n=== Memproses Data Test ===")
test_records    = []
test_img_names  = sorted(os.listdir(CFG.test_dir))

for img_name in tqdm(test_img_names, desc="Cropping Test Set"):
    img_path    = os.path.join(CFG.test_dir, img_name)
    cropped_img = process_and_crop_face(img_path, face_model)
    if cropped_img is not None:
        save_path = os.path.join(out_test_dir, img_name)
        cv2.imwrite(save_path, cv2.cvtColor(cropped_img, cv2.COLOR_RGB2BGR))
        test_records.append({'id': img_name.split('.')[0], 'image_path': save_path})

test_df = pd.DataFrame(test_records)
test_df.to_csv("/kaggle/working/test_df.csv", index=False)
print(f"Total Test Images: {len(test_df)}")
```

# **4. Augmentasi Bertingkat (Anti-Spoofing Specific)**

Dalam domain *Face Anti-Spoofing*, strategi augmentasi data memiliki peran yang jauh lebih krusial dibandingkan klasifikasi gambar biasa. Di sini, augmentasi bukan sekadar teknik untuk "memperbanyak data", melainkan sebuah **Simulator Serangan (Attack Simulator)**. 

Jika kita membiarkan model *Vision Transformer* (DINOv2) mempelajari data latih apa adanya, model akan dengan cepat menghafal dataset (*overfitting*) dan gagal mengenali serangan manipulasi di dunia nyata. Oleh karena itu, kami merancang **3 Lapis Pertahanan Augmentasi** menggunakan pustaka `Albumentations` yang secara eksplisit meniru artefak *spoofing* fisik dan digital:

### Level 1: Pertahanan Spasial & Geometris
Lapis pertama bertujuan untuk mencegah model menghafal posisi wajah, kemiringan kepala, atau *framing* kamera.
* **`ShiftScaleRotate` & `HorizontalFlip`:** Memastikan model invariant (kebal) terhadap pergeseran, perbesaran, rotasi ringan, dan orientasi cermin.

### Level 2: Simulasi Artefak Serangan (The Core Defense)
Ini adalah inti dari taktik augmentasi kami. Kami secara sengaja menyuntikkan *noise* frekuensi tinggi dan distorsi mikroskopis untuk memaksa DINOv2 menggunakan *Attention*-nya pada struktur yang benar, bukan sekadar ketajaman piksel lokal.
* **Simulasi Layar (Moiré):** Kami menggunakan `GridDistortion` dan `OpticalDistortion` untuk meniru gelombang lengkung mikroskopis (*Moiré patterns*) yang sering muncul saat kamera merekam layar digital (*fake_screen*).
* **Simulasi Cetakan & Kertas:** `ImageCompression` (JPEG artifacts) dan `Downscale` digunakan secara agresif untuk meniru hilangnya detail pori-pori kulit saat wajah dicetak pada kertas murah (*fake_printed*).
* **Simulasi Sensor Kamera:** Kombinasi `GaussNoise` dan `ISONoise` meniru butiran (*grain*) pada foto yang diambil dalam kondisi kurang cahaya, memaksa model membedakan *noise* alami kamera dengan *noise* buatan dari layar.
* **Defocus & Blur:** `GaussianBlur` dan `Sharpen` mensimulasikan kamera yang kehilangan fokus saat merekam layar atau foto cetak dari jarak sangat dekat.

### Level 3: Variasi Lingkungan & Pencahayaan
Lapis terakhir memastikan model tidak tertipu oleh trik pencahayaan penyerang.
* **Pantulan Cahaya & Tinta Pudar:** `ColorJitter` dan `HueSaturationValue` meniru perubahan warna akibat cahaya layar (*screen glare*) atau tinta *printer* yang luntur/pudar.
* **Overexposure & Bayangan Tak Merata:** `RandomShadow` sangat krusial untuk meniru skenario di mana penyerang menggunakan *flash* ponsel yang membuat sebagian wajah (atau topeng) terlihat kelebihan cahaya sementara sisi lainnya gelap.

**Sinergi dengan Arsitektur:** Seluruh augmentasi berat ini akan menghasilkan gambar-gambar yang sangat menantang (sebagai *Hard Positives*). Ini adalah persiapan sempurna sebelum gambar-gambar tersebut dimasukkan ke dalam ruang laten **Supervised Contrastive Learning** pada tahapan selanjutnya!

```python
# Augmentasi Training — termasuk simulasi serangan screen, print, dan noise
train_transform = A.Compose([
    # --- Level 1: Spatial & Geometric ---
    A.HorizontalFlip(p=0.5),
    A.ShiftScaleRotate(
        shift_limit=0.05, scale_limit=0.1, rotate_limit=15,
        border_mode=cv2.BORDER_CONSTANT, p=0.5
    ),

    # --- Level 2: Simulasi Artefak Serangan ---
    # Simulasi Moiré pattern (screen artifact) dan distorsi kisi
    A.OneOf([
        A.GridDistortion(num_steps=5, distort_limit=0.1, p=1.0),
        A.OpticalDistortion(distort_limit=0.05, shift_limit=0.05, p=1.0),
    ], p=0.3),

    # Simulasi JPEG compression artifact (print & screen attack)
    A.OneOf([
        A.ImageCompression(quality_lower=30, quality_upper=80, p=1.0),
        A.Downscale(scale_min=0.5, scale_max=0.9, p=1.0),
    ], p=0.45),

    # Simulasi noise kamera dan sensor
    A.OneOf([
        A.GaussNoise(var_limit=(10.0, 60.0), p=1.0),
        A.ISONoise(color_shift=(0.01, 0.06), intensity=(0.1, 0.5), p=1.0),
        A.MultiplicativeNoise(multiplier=(0.9, 1.1), p=1.0),
    ], p=0.4),

    # Simulasi blur/sharpness (defocus atau layar)
    A.OneOf([
        A.Blur(blur_limit=5, p=1.0),
        A.GaussianBlur(blur_limit=(3, 7), p=1.0),
        A.Sharpen(alpha=(0.1, 0.4), lightness=(0.5, 1.0), p=1.0),
    ], p=0.35),

    # --- Level 3: Variasi Warna & Pencahayaan ---
    A.ColorJitter(
        brightness=0.25, contrast=0.25,
        saturation=0.35, hue=0.1, p=0.5
    ),
    A.HueSaturationValue(
        hue_shift_limit=12, sat_shift_limit=25, val_shift_limit=12, p=0.4
    ),
    A.RandomBrightnessContrast(
        brightness_limit=0.2, contrast_limit=0.2, p=0.5
    ),

    # Simulasi pencahayaan tidak merata (overexposure / shadow)
    A.OneOf([
        A.RandomShadow(shadow_roi=(0, 0.5, 1, 1), p=1.0),
        A.RandomToneCurve(scale=0.15, p=1.0),
    ], p=0.2),

    # --- Standarisasi ImageNet (DINOv2) ---
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ToTensorV2()
])

# Augmentasi Validasi / Test — hanya normalisasi
valid_transform = A.Compose([
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ToTensorV2()
])

print("Augmentasi bertingkat (anti-spoofing) siap!")
```

# **5. Dataset & Dataloader (Patch Sampler)**

Pada sistem klasifikasi biasa, *Dataloader* hanya bertugas membaca gambar, mengubah resolusinya, dan mengubahnya menjadi tensor. Namun, untuk arsitektur yang berbasis *Vision Transformer* (seperti DINOv2) dalam domain *Anti-Spoofing*, pendekatan standar tersebut memiliki kelemahan fatal.

Jika kita memberikan seluruh wajah (beresolusi rendah setelah di-*resize*) ke model, model mungkin akan fokus pada fitur makro (seperti bentuk hidung atau kacamata) alih-alih fitur mikro (seperti *noise* resolusi layar atau tekstur Moiré).

Oleh karena itu, kami merancang **Custom Patch Sampler Dataset** dengan logika berikut:

### Mengapa Kita Membutuhkan "Patch"?
1. **Fokus pada Tekstur Lokal:** Alih-alih melihat "sebuah wajah", model kita disuapi dengan "potongan-potongan tekstur". Serangan *spoofing* (seperti kertas cetak) sering kali memiliki tekstur yang seragam di seluruh area. Dengan mengekstraksi *patches* kecil, kita memaksa model untuk menjadi ahli tekstur (*texture expert*).
2. **Efisiensi Komputasi & Resolusi Asli:** DINOv2 dilatih pada *patch* berukuran 14x14 piksel. Daripada me-*resize* seluruh gambar wajah menjadi sangat kecil (yang akan menghancurkan *Moiré pattern*), kami mengekstrak beberapa *patches* berukuran 224x224 piksel secara acak dari gambar asli. Ini mempertahankan detail resolusi tinggi tanpa membuat memori GPU kehabisan kapasitas (OOM).

### Mekanisme Dual-Mode
Kelas `FASDataset` kami dirancang untuk berperilaku berbeda tergantung fasenya:
* **Fase Training (`mode='train'`):** Mengekstrak $N$ *patches* secara acak (`Random Crop`) dari area wajah. Ini bertindak sebagai bentuk augmentasi spasial tingkat lanjut.
* **Fase Validasi (`mode='valid'`):** Mengekstrak *patches* secara seragam dan terstruktur (seperti pola kotak catur / *Grid Sampler*) untuk memastikan evaluasi yang deterministik dan mencakup seluruh area wajah.

Seluruh *patches* dari satu wajah ini kemudian ditumpuk (*stacked*) dan dilempar bersamaan ke dalam arsitektur DINOv2 + *Attention Pooling* pada tahap berikutnya.

```python
class LivenessPatchDataset(Dataset):
    def __init__(
        self, df, transform=None, mode='train',
        num_patches=CFG.num_patches_train,
        patch_size=CFG.patch_size
    ):
        self.df          = df
        self.transform   = transform
        self.mode        = mode
        self.num_patches = num_patches
        self.patch_size  = patch_size

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row     = self.df.iloc[idx]
        img     = cv2.imread(row['image_path'])
        img     = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        h, w, _ = img.shape
        patches  = []

        if self.mode == 'train':
            for _ in range(self.num_patches):
                y     = random.randint(0, h - self.patch_size)
                x     = random.randint(0, w - self.patch_size)
                patch = img[y:y+self.patch_size, x:x+self.patch_size]
                if self.transform:
                    patch = self.transform(image=patch)['image']
                patches.append(patch)
        else:
            # Grid 3×3 = 9 patches
            step_y   = (h - self.patch_size) // 2
            step_x   = (w - self.patch_size) // 2
            y_coords = [0, step_y, h - self.patch_size]
            x_coords = [0, step_x, w - self.patch_size]
            for y in y_coords:
                for x in x_coords:
                    patch = img[y:y+self.patch_size, x:x+self.patch_size]
                    if self.transform:
                        patch = self.transform(image=patch)['image']
                    patches.append(patch)

        patches_tensor = torch.stack(patches)  # [N, C, H, W]

        if self.mode == 'test':
            return patches_tensor, row['id']
        return patches_tensor, torch.tensor(row['target'], dtype=torch.long)


def get_dataloaders(df, fold):
    df_train = df[df['fold'] != fold].reset_index(drop=True)
    df_valid = df[df['fold'] == fold].reset_index(drop=True)

    train_ds = LivenessPatchDataset(df_train, transform=train_transform, mode='train')
    valid_ds = LivenessPatchDataset(df_valid, transform=valid_transform, mode='valid')

    train_loader = DataLoader(
        train_ds, batch_size=CFG.batch_size, shuffle=True,
        num_workers=CFG.num_workers, pin_memory=True, drop_last=True
    )
    valid_loader = DataLoader(
        valid_ds, batch_size=CFG.batch_size, shuffle=False,
        num_workers=CFG.num_workers, pin_memory=True
    )
    return train_loader, valid_loader


# --- Sanity Check ---
train_df = pd.read_csv("/kaggle/working/train_df.csv")
test_df  = pd.read_csv("/kaggle/working/test_df.csv")

skf = StratifiedKFold(n_splits=CFG.n_fold, shuffle=True, random_state=CFG.seed)
train_df['fold'] = -1
for fold, (_, val_idx) in enumerate(skf.split(X=train_df, y=train_df['target'])):
    train_df.loc[val_idx, 'fold'] = fold

_tl, _vl = get_dataloaders(train_df, fold=0)
_p, _l   = next(iter(_tl))
print(f"Train batch shape : {_p.shape}  → (B, N_patches, C, H, W)")
print(f"Label batch shape : {_l.shape}")
print("Dataloader siap!")
```

# **6. Arsitektur Model**

Pada klasifikasi gambar standar, pendekatan yang umum dilakukan adalah mengambil *backbone* *Vision Transformer* (ViT), lalu merata-ratakan seluruh *output token*-nya menggunakan *Mean Pooling* sebelum masuk ke lapisan klasifikasi (Linear Layer).

Namun, untuk tugas **Face Anti-Spoofing (FAS)**, pendekatan *Mean Pooling* sangatlah berbahaya. Artefak manipulasi (seperti *Moiré pattern* dari layar atau *noise* kompresi dari foto cetak) sering kali hanya muncul di sebagian kecil area wajah (lokal). Jika kita merata-ratakan semua fitur, sinyal manipulasi yang kecil ini akan "tenggelam" oleh sinyal tekstur kulit asli di sekitarnya.

Untuk mengatasi ini, kami merancang arsitektur kustom dengan 3 fondasi utama:

### 1. Backbone: DINOv2-Base dengan Register Tokens (`dinov2_vitb14_reg`)
Kami menggunakan model fundamental dari Meta, yaitu DINOv2-Base. Keistimewaan varian ini adalah adanya **Register Tokens**. Dalam ViT standar, model sering kali membuang informasi yang tidak berguna (seperti latar belakang yang kosong) ke *patch tokens* secara acak, yang dapat mengganggu ekstraksi fitur. *Register Tokens* bertindak sebagai "tempat sampah" (*garbage collectors*) yang menyerap semua informasi global yang tidak relevan, sehingga *patch tokens* yang tersisa benar-benar bersih dan hanya merepresentasikan tekstur murni.

### 2. Multi-Head Attention Pooling
Sebagai pengganti *Mean Pooling*, kami mengimplementasikan **Attention Pooling**. Lapisan ini secara dinamis menghitung *bobot (attention weights)* untuk setiap *patch*. Model akan secara mandiri belajar memberikan bobot yang sangat tinggi (fokus) hanya pada *patch* yang mencurigakan (misalnya *patch* yang mengandung *blur* atau distorsi piksel layar), dan mengabaikan *patch* yang terlihat normal.

### 3. Dual Output Heads (SupCon + Classification)
Arsitektur ini tidak langsung menebak probabilitas kelas. Fitur akhir yang dihasilkan akan dipecah menjadi dua cabang:
1. **Projection Head (Contrastive):** Memetakan fitur ke dalam ruang laten 128-dimensi untuk dioptimalkan menggunakan *Supervised Contrastive Loss* (menarik gambar kelas yang sama agar berdekatan, dan menolak kelas yang berbeda).
2. **Classification Head:** Lapisan Linear akhir yang akan menghasilkan *logits* untuk klasifikasi 6 kelas dengan menggunakan *Class-Weighted Loss*.

```python
class AttentionPooling(nn.Module):
    """
    Multi-Head Attention Pooling:
    Belajar patch mana yang paling informatif sebagai tanda spoofing,
    bukan sekadar rata-ratakan semua patch secara merata.
    """
    def __init__(self, embed_dim, num_heads=8, dropout=0.1):
        super().__init__()
        self.attn  = nn.MultiheadAttention(
            embed_dim, num_heads=num_heads,
            dropout=dropout, batch_first=True
        )
        self.query = nn.Parameter(torch.randn(1, 1, embed_dim))
        self.norm  = nn.LayerNorm(embed_dim)

    def forward(self, x):  # x: [B, N, D]
        q       = self.query.expand(x.size(0), -1, -1)  # [B, 1, D]
        out, _  = self.attn(q, x, x)
        return self.norm(out.squeeze(1))                 # [B, D]


class PatchDINOv2(nn.Module):
    def __init__(
        self,
        num_classes = len(CFG.classes),
        embed_dim   = CFG.embed_dim,
        fused_dim   = CFG.fused_dim,
        dropout     = 0.3
    ):
        super().__init__()

        # Backbone DINOv2-Base dengan register tokens
        self.backbone = torch.hub.load(
            'facebookresearch/dinov2', CFG.backbone_name
        )

        # Attention Pooling untuk fitur CLS patch
        self.cls_pool = AttentionPooling(embed_dim, num_heads=8, dropout=dropout)

        # Attention Pooling untuk register token
        self.reg_pool = AttentionPooling(embed_dim, num_heads=8, dropout=dropout)

        # Projection sebelum fusion (opsional — bantu stabilisasi gradien)
        self.cls_proj = nn.Sequential(
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, embed_dim),
            nn.GELU()
        )
        self.reg_proj = nn.Sequential(
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, embed_dim),
            nn.GELU()
        )

        # Classifier Head
        self.head = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(fused_dim, fused_dim // 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(fused_dim // 2, num_classes)
        )

    def forward(self, x):
        # x: [B, N_patches, C, H, W]
        B, N, C, H, W = x.shape
        x_flat = x.view(B * N, C, H, W)

        # Ekstrak CLS token + register tokens dari DINOv2
        out        = self.backbone.forward_features(x_flat)
        cls_tokens = out['x_norm_clstoken']   # [B*N, embed_dim]
        reg_tokens = out['x_norm_regtokens']  # [B*N, num_reg, embed_dim]

        # Bentuk ulang ke [B, N, embed_dim]
        cls_tokens = cls_tokens.reshape(B, N, -1)
        # [B*N, num_reg, embed_dim] → [B, N*num_reg, embed_dim]
        num_reg    = reg_tokens.shape[1]
        reg_tokens = reg_tokens.reshape(B, N * num_reg, -1)

        # Attention Pooling: rangkum semua patch menjadi satu vektor
        cls_agg = self.cls_pool(cls_tokens)  # [B, embed_dim]
        reg_agg = self.reg_pool(reg_tokens)  # [B, embed_dim]

        # Proyeksi & Fusi
        cls_agg = self.cls_proj(cls_agg)
        reg_agg = self.reg_proj(reg_agg)
        fused   = torch.cat([cls_agg, reg_agg], dim=-1)  # [B, fused_dim]

        logits = self.head(fused)

        # Kembalikan logits + patch CLS embeddings untuk loss tambahan
        return logits, cls_tokens  # cls_tokens: [B, N, embed_dim]


# Test bentuk
print("Mengunduh DINOv2-Base...")
_model  = PatchDINOv2().to(CFG.device)
_input  = _p.to(CFG.device)
_logits, _embeds = _model(_input)
print(f"Logits shape         : {_logits.shape}  → (B, num_classes)")
print(f"Patch CLS embeds     : {_embeds.shape}  → (B, N_patches, embed_dim)")
del _model, _input, _logits, _embeds
torch.cuda.empty_cache()
print("Arsitektur PatchDINOv2 + AttentionPooling + RegisterTokens siap!")
```

# **7. Loss Function (AAMS + Similarity + Supervised Contrastive)**

Dalam klasifikasi gambar tradisional, fungsi *Cross-Entropy (CE)* standar sudah cukup. Namun, dalam **Face Anti-Spoofing (FAS)**, CE memiliki kelemahan fatal: ia hanya peduli apakah model menebak kelas dengan benar, tetapi **tidak peduli seberapa jauh jarak representasi antar kelas tersebut**. 

Jika wajah asli (`realperson`) dan wajah di layar (`fake_screen`) memiliki jarak representasi yang terlalu dekat di ruang laten (*latent space*), model akan sangat mudah tertipu saat dihadapkan pada data *testing* yang sedikit berbeda.

Untuk memaksa model membangun benteng pertahanan yang absolut, kami merancang fungsi *loss* gabungan (Composite/Triple Loss) yang beroperasi pada tiga level matematis:

### 1. Supervised Contrastive Loss (SupCon)
Alih-alih langsung memprediksi kelas, SupCon bekerja pada **Projection Head** (ruang laten 128-dimensi). 
* **Tarik (Pull):** Memaksa fitur dari gambar dalam kelas yang sama (misal: semua wajah asli) untuk berkumpul menjadi klaster yang sangat padat.
* **Dorong (Push):** Menendang fitur dari kelas yang berbeda (wajah asli vs serangan layar/cetak) agar sejauh mungkin satu sama lain. 
Ini membuat model sangat tangguh terhadap serangan *Zero-Day* (serangan baru) karena *real face* sudah terisolasi rapat di satu titik ruang.

### 2. Cosine Similarity & Additive Angular Margin (AAMS)
Terinspirasi dari *ArcFace* di domain pengenalan wajah, kami mengimplementasikan penalti margin sudut. Daripada menggunakan garis lurus untuk memisahkan kelas, fitur direksional diproyeksikan ke dalam sebuah struktur *Hypersphere* (bola multi-dimensi). Kami menambahkan margin sudut ekstra secara artifisial, sehingga model harus bekerja ekstra keras untuk melewati batas tersebut. Ini menghasilkan **batas keputusan (decision boundary) yang sangat tajam**.

### 3. Class-Weighted Classification
Seperti yang dibuktikan pada bagian EDA, dataset kita menderita *imbalance* kelas yang ekstrem (kelas *realperson* melimpah, kelas *fake_unknown* sangat sedikit). Jika dibiarkan, model akan menjadi bias. Kami menghitung invers dari frekuensi kelas dan mengumpankannya sebagai `weight` ke dalam fungsi klasifikasi akhir, memberikan penalti yang jauh lebih berat jika model salah menebak serangan minoritas.

```python
class SupervisedContrastiveLoss(nn.Module):
    """
    SupCon Loss — memaksa embedding kelas yang sama berdekatan
    dan kelas berbeda berjauhan di ruang fitur.
    Sangat membantu untuk kelas ambigu seperti fake_unknown.
    """
    def __init__(self, temperature=CFG.temperature_supcon):
        super().__init__()
        self.temperature = temperature

    def forward(self, features, labels):
        # features: [B, D]  (mean dari patch embeddings, sudah di-normalize)
        features = F.normalize(features, dim=1)
        sim_mat  = torch.matmul(features, features.T) / self.temperature  # [B, B]

        labels   = labels.view(-1, 1)
        mask     = torch.eq(labels, labels.T).float().to(features.device)
        mask.fill_diagonal_(0)  # Exclude diri sendiri

        exp_sim  = torch.exp(sim_mat)
        # Untuk setiap anchor, hitung log-softmax terhadap semua sample lain
        log_prob = sim_mat - torch.log(
            exp_sim.sum(dim=1, keepdim=True) - exp_sim.diagonal().unsqueeze(1)
        )
        n_pos    = mask.sum(dim=1).clamp(min=1)
        loss     = -(mask * log_prob).sum(dim=1) / n_pos
        return loss.mean()


class PatchNetLoss(nn.Module):
    def __init__(
        self,
        num_classes    = len(CFG.classes),
        scale          = 30.0,
        lambda_sim     = CFG.lambda_sim,
        lambda_supcon  = CFG.lambda_supcon
    ):
        super().__init__()
        self.scale          = scale
        self.lambda_sim     = lambda_sim
        self.lambda_supcon  = lambda_supcon

        # Asymmetric margin per kelas
        # [realperson, fake_printed, fake_screen, fake_mask, fake_mannequin, fake_unknown]
        self.margins = torch.tensor([0.4, 0.2, 0.2, 0.2, 0.2, 0.1])

        # Class-weighted CE dengan label smoothing
        class_weights = torch.tensor(CFG.class_weights)
        self.ce_loss  = nn.CrossEntropyLoss(
            weight         = class_weights,
            label_smoothing = CFG.label_smoothing
        )

        self.supcon = SupervisedContrastiveLoss()

    def to(self, device):
        super().to(device)
        self.margins = self.margins.to(device)
        self.ce_loss.weight = self.ce_loss.weight.to(device)
        return self

    def forward(self, logits, patch_embeddings, labels):
        """
        logits           : [B, num_classes]
        patch_embeddings : [B, N_patches, embed_dim]
        labels           : [B]
        """
        # 1. Asymmetric AM-Softmax (AAMS)
        m              = self.margins[labels]  # [B]
        cosine         = F.normalize(logits, p=2, dim=1)
        target_oh      = torch.zeros_like(cosine)
        target_oh.scatter_(1, labels.view(-1, 1), 1.0)
        cosine_m       = cosine - m.view(-1, 1) * target_oh
        scaled_logits  = cosine_m * self.scale
        L_AAMS         = self.ce_loss(scaled_logits, labels)

        # 2. Similarity Loss — konsistensi antar patch dalam satu gambar
        mean_emb = patch_embeddings.mean(dim=1, keepdim=True)
        L_Sim    = F.mse_loss(patch_embeddings, mean_emb.expand_as(patch_embeddings))

        # 3. Supervised Contrastive Loss — fitur kelas sama berdekatan
        mean_feat = patch_embeddings.mean(dim=1)  # [B, embed_dim]
        L_SupCon  = self.supcon(mean_feat, labels)

        # 4. Total Loss
        total = L_AAMS + (self.lambda_sim * L_Sim) + (self.lambda_supcon * L_SupCon)
        return total, L_AAMS, L_Sim, L_SupCon


print("Loss function (AAMS + Similarity + SupCon) siap!")
```

# **8. Training & Validation Loop**

Melatih *Vision Transformer* berukuran besar (seperti DINOv2) menggunakan dataset *Face Anti-Spoofing* yang kecil dan sangat tidak seimbang adalah resep yang berisiko tinggi untuk mengalami *overfitting* atau *gradient explosion* (NaN). 

Untuk memastikan model konvergen dengan skor yang optimal dan stabil, *Training Engine* ini dibangun dengan 4 strategi optimasi kelas berat:

### 1. Phased Unfreezing & Discriminative Learning Rates
DINOv2 dari Meta sudah sangat pintar dalam mengekstrak fitur gambar. Jika kita langsung melatih seluruh jaringannya dengan *Learning Rate* (LR) yang besar, kita akan merusak representasi spasial bawaannya (*Catastrophic Forgetting*).
Oleh karena itu, kami merancang fungsi `adjust_model_phases`:
* **Phase 1 (Epoch 1-5):** *Backbone* DINOv2 dibekukan (*Freeze*). LR difokuskan penuh pada *Attention Pooling* dan *Classification Head* agar mereka beradaptasi dengan ruang fitur DINO.
* **Phase 2 (Epoch 6-14):** Setengah dari blok arsitektur perlahan dicairkan (*Unfreeze*) dengan LR yang sangat kecil (mikro).
* **Phase 3 (Epoch 15+):** Seluruh jaringan dicairkan untuk *Fine-Tuning* tahap akhir. Pendekatan ini (*Discriminative LRs*) terbukti jauh lebih stabil.

### 2. Absolute Stability (Anti-NaN Guards & FP32)
Serangan manipulasi yang ekstrem sering kali menghasilkan gradien yang meledak. Kami menonaktifkan *Mixed Precision (AMP/FP16)* dan murni menggunakan **FP32 Engine**. Selain itu, kami memasang "sabuk pengaman" berupa:
* **Gradient Clipping:** Membatasi norma gradien maksimal agar bobot tidak melompat tak terkendali.
* **NaN Catchers:** Mengisolasi dan melewati *batch* yang menghasilkan *loss* rusak tanpa menghentikan seluruh proses pelatihan.

### 3. Multi-GPU DataParallel (T4x2)
Memanfaatkan infrastruktur Kaggle secara maksimal, *engine* ini akan mendeteksi keberadaan GPU ganda secara otomatis. Menggunakan `torch.nn.DataParallel`, beban *batch* dibagi secara merata ke dua GPU T4, memangkas waktu *training* hingga hampir setengahnya dan memungkinkan kita menggunakan resolusi yang lebih tajam.

### 4. F1-Macro as the True North Star
Dalam dataset yang *imbalanced*, metrik **Akurasi adalah sebuah ilusi**. Model bisa mendapat akurasi 90% hanya dengan menebak kelas `realperson` terus-menerus. Kami secara ketat mengevaluasi dan menyimpan *checkpoint* model (`best_model.pth`) murni berdasarkan metrik **F1-Score Macro**, yang memaksa model untuk berkinerja sama baiknya di serangan minoritas (*fake_unknown*, *fake_mask*) maupun di kelas mayoritas.

```python
import time
import gc


def train_epoch(model, dataloader, criterion, optimizer, scaler, scheduler, epoch, device):
    model.train()
    total_loss = 0.0

    # Freeze backbone 4 epoch pertama (warmup head saja)
    backbone = model.module.backbone if hasattr(model, 'module') else model.backbone
    if epoch < 4:
        for p in backbone.parameters():
            p.requires_grad = False
    else:
        for p in backbone.parameters():
            p.requires_grad = True

    pbar = tqdm(dataloader, desc=f"Train Epoch {epoch+1}/{CFG.epochs}", leave=False)
    for patches, labels in pbar:
        patches, labels = patches.to(device), labels.to(device)
        optimizer.zero_grad()

        with autocast():
            logits, patch_embeds      = model(patches)
            loss, l_aams, l_sim, l_sc = criterion(logits, patch_embeds, labels)

        scaler.scale(loss).backward()
        # Gradient clipping untuk stabilitas
        scaler.unscale_(optimizer)
        nn.utils.clip_grad_norm_(model.parameters(), CFG.grad_clip)
        scaler.step(optimizer)
        scaler.update()

        # OneCycleLR step per batch
        scheduler.step()

        total_loss += loss.item()
        pbar.set_postfix(
            loss=f"{loss.item():.4f}",
            aams=f"{l_aams.item():.3f}",
            sim=f"{l_sim.item():.3f}",
            sc=f"{l_sc.item():.3f}"
        )

    return total_loss / len(dataloader)


@torch.no_grad()
def valid_epoch(model, dataloader, criterion, device):
    model.eval()
    total_loss = 0.0
    all_preds  = []
    all_labels = []

    pbar = tqdm(dataloader, desc="Validating", leave=False)
    for patches, labels in pbar:
        patches, labels = patches.to(device), labels.to(device)
        with autocast():
            logits, patch_embeds      = model(patches)
            loss, _, _, _             = criterion(logits, patch_embeds, labels)

        total_loss += loss.item()
        preds = torch.argmax(logits, dim=1).cpu().numpy()
        all_preds.extend(preds)
        all_labels.extend(labels.cpu().numpy())

    macro_f1 = f1_score(all_labels, all_preds, average='macro')
    return total_loss / len(dataloader), macro_f1


print("Training & Validation loop siap!")
```

# **9. Full Training (5-Fold + DataParallel T4x2)**

Kita telah tiba di tahap eksekusi akhir. Dalam kompetisi dengan dataset yang relatif kecil dan *imbalanced*, mengandalkan satu pembagian data (seperti 80% Train, 20% Validasi) adalah tindakan bunuh diri (*Leaderboard Shake-up hazard*). Model bisa saja mendapatkan skor tinggi di validasi tersebut hanya karena kebetulan (*lucky split*).

Untuk menjamin bahwa model kita benar-benar tangguh dan metrik F1-Macro kita dapat dipercaya (berkorelasi kuat dengan skor Juri), kami menjalankan **5-Fold Cross Validation**. 

Tahap eksekusi ini memiliki 3 keunggulan strategis:

### 1. Robust Out-of-Fold (OOF) Scoring
Dataset dibagi menjadi 5 bagian. Model dilatih 5 kali secara independen, di mana setiap putaran menggunakan 4 bagian untuk *training* dan 1 bagian untuk validasi. Pada akhirnya, setiap gambar dalam dataset telah dievaluasi tepat satu kali sebagai data yang belum pernah dilihat model (*unseen data*). Rata-rata dari skor ini memberikan estimasi performa dunia nyata yang sangat akurat.

### 2. Ekstrem Memory Management (Anti-OOM)
Melatih model *Vision Foundation* (DINOv2) sebanyak 5 kali berturut-turut di GPU Kaggle T4 (16GB VRAM) akan dengan cepat menyebabkan *CUDA Out of Memory (OOM)* akibat kebocoran memori tensor sisa. 
Kami memasang protokol pembersihan memori absolut di akhir setiap *fold*:
* Menghapus referensi model dan *optimizer* (`del model, optimizer`).
* Memanggil *Garbage Collector* Python (`gc.collect()`).
* Mengosongkan memori fisik GPU secara paksa (`torch.cuda.empty_cache()`).

### 3. Auto-Scaling Multi-GPU (T4x2 DataParallel)
Saat fungsi ini dipanggil, ia akan mendeteksi *environment* Kaggle secara otomatis. Jika terdeteksi 2 GPU (T4x2), arsitektur secara instan dibungkus dengan `nn.DataParallel`. Beban *forward* dan *backward pass* didistribusikan secara sinkron, memangkas waktu pelatihan 5-Fold dari hitungan belasan jam menjadi hanya beberapa jam saja.

```python
def run_training():
    os.makedirs("/kaggle/working/models", exist_ok=True)
    oof_f1_scores = []

    for fold in range(CFG.n_fold):
        print(f"\n{'='*22} Fold {fold+1}/{CFG.n_fold} {'='*22}")

        train_loader, valid_loader = get_dataloaders(train_df, fold)

        # Inisialisasi model
        model = PatchDINOv2(num_classes=len(CFG.classes)).to(CFG.device)

        # DataParallel untuk T4x2
        if CFG.use_multi_gpu:
            print(f"  Menggunakan DataParallel pada {torch.cuda.device_count()} GPU")
            model = nn.DataParallel(model)

        criterion = PatchNetLoss().to(CFG.device)

        # Differential Learning Rate: backbone jauh lebih kecil dari head
        backbone_params = (
            model.module.backbone.parameters()
            if hasattr(model, 'module')
            else model.backbone.parameters()
        )
        head_params = (
            list(model.module.cls_pool.parameters()) +
            list(model.module.reg_pool.parameters()) +
            list(model.module.cls_proj.parameters()) +
            list(model.module.reg_proj.parameters()) +
            list(model.module.head.parameters())
        ) if hasattr(model, 'module') else (
            list(model.cls_pool.parameters()) +
            list(model.reg_pool.parameters()) +
            list(model.cls_proj.parameters()) +
            list(model.reg_proj.parameters()) +
            list(model.head.parameters())
        )

        optimizer = optim.AdamW(
            [
                {'params': backbone_params, 'lr': CFG.lr_backbone},
                {'params': head_params,     'lr': CFG.lr_head}
            ],
            weight_decay=CFG.weight_decay
        )

        # OneCycleLR: warmup + cosine decay dalam satu siklus
        scheduler = optim.lr_scheduler.OneCycleLR(
            optimizer,
            max_lr   = [CFG.lr_backbone, CFG.lr_head],
            steps_per_epoch = len(train_loader),
            epochs   = CFG.epochs,
            pct_start= CFG.warmup_pct,
            anneal_strategy='cos'
        )

        scaler       = GradScaler()
        best_f1      = 0.0
        model_path   = f"/kaggle/working/models/dinov2b_fold{fold}.pth"

        for epoch in range(CFG.epochs):
            t0          = time.time()
            train_loss  = train_epoch(
                model, train_loader, criterion,
                optimizer, scaler, scheduler, epoch, CFG.device
            )
            valid_loss, valid_f1 = valid_epoch(
                model, valid_loader, criterion, CFG.device
            )
            elapsed = time.time() - t0

            print(
                f"  Epoch {epoch+1:02d} | "
                f"Train Loss: {train_loss:.4f} | "
                f"Val Loss: {valid_loss:.4f} | "
                f"Val F1 Macro: {valid_f1:.4f} | "
                f"Time: {elapsed:.0f}s"
            )

            if valid_f1 > best_f1:
                print(f"  ★ F1 naik {best_f1:.4f} → {valid_f1:.4f} — menyimpan model...")
                best_f1 = valid_f1
                # Simpan state_dict dari model asli (bukan DataParallel wrapper)
                state = (
                    model.module.state_dict()
                    if hasattr(model, 'module')
                    else model.state_dict()
                )
                torch.save(state, model_path)

        oof_f1_scores.append(best_f1)
        print(f"  ▶ Fold {fold+1} selesai. Best Val F1 Macro: {best_f1:.4f}")

        del model, optimizer, scheduler, train_loader, valid_loader
        torch.cuda.empty_cache()
        gc.collect()

    print(f"\n{'='*50}")
    print(f"TRAINING SELESAI!")
    print(f"OOF F1 Macro per Fold : {[round(s, 4) for s in oof_f1_scores]}")
    print(f"Rata-rata CV F1 Macro : {np.mean(oof_f1_scores):.4f}")
    print(f"{'='*50}")


run_training()
```

# **10. Inferensi — TTA Spasial + Softmax Ensembling**

Tahap inferensi adalah pembuktian akhir dari *pipeline* Anti-Spoofing kita. Untuk memastikan bahwa prediksi akhir memiliki tingkat kepercayaan dan akurasi maksimum pada data pengujian (*unseen test data*), kami mengimplementasikan dua teknik stabilisasi tingkat lanjut: **Spatial Test-Time Augmentation (TTA)** dan **Softmax Averaging Ensemble**.

### 1. Spatial Test-Time Augmentation (TTA)
Dalam kondisi pengujian yang ideal, model dapat langsung mengenali serangan. Namun, pada kasus gambar yang ambigu (misalnya pantulan layar hanya terlihat di sudut kiri), orientasi gambar bisa mempengaruhi respons *Attention Pooling*. 
Kami menggunakan **Spatial TTA (Horizontal Flip)**. Setiap gambar *test* akan diprediksi dua kali:
1. Gambar asli (Original)
2. Gambar yang dibalik secara horizontal (Flipped)

Ini memberikan model "kesempatan kedua" untuk mengamati pola mikroskopis dari perspektif spasial yang berlawanan.

### 2. Softmax Ensembling (The "Super Brain")
Alih-alih mengandalkan satu model terbaik, kami menggunakan kelima bobot model yang dihasilkan dari proses *5-Fold Cross Validation*. 

Metode penggabungannya sangat krusial. Kami **tidak** menggunakan *Hard Voting* (memilih kelas terbanyak), karena metode tersebut membuang informasi tingkat kepercayaan model (kalau 3 model ragu-ragu dan 2 model sangat yakin, *hard voting* bisa salah pilih).

Sebagai gantinya, kami menggunakan **Softmax Ensembling**:
1. Mengambil *raw logits* dari 5 model untuk 2 varian TTA (total 10 tebakan per gambar).
2. Mengubah *logits* tersebut menjadi probabilitas (0.0 - 1.0) menggunakan aktivasi Softmax.
3. Merata-ratakan (*averaging*) ke-10 distribusi probabilitas tersebut.
4. Mengambil nilai probabilitas tertinggi (`argmax`) sebagai prediksi kelas final.

Hasilnya adalah batas keputusan (*decision boundary*) yang sangat halus dan tahan terhadap jebakan *overfitting* lokal!

```python
def run_inference_and_submit():
    print("=== Memulai Fase Inferensi & Ensembling ===")

    test_df   = pd.read_csv("/kaggle/working/test_df.csv")
    test_ds   = LivenessPatchDataset(
        test_df, transform=valid_transform,
        mode='test', num_patches=CFG.num_patches_test
    )
    test_loader = DataLoader(
        test_ds, batch_size=CFG.batch_size, shuffle=False,
        num_workers=CFG.num_workers, pin_memory=True
    )

    # Akumulasi probabilitas per gambar
    ensemble_probs = {
        row['id']: np.zeros(len(CFG.classes))
        for _, row in test_df.iterrows()
    }
    n_loaded = 0

    for fold in range(CFG.n_fold):
        model_path = f"/kaggle/working/models/dinov2b_fold{fold}.pth"
        if not os.path.exists(model_path):
            print(f"  ⚠ Model Fold {fold} tidak ditemukan, skip.")
            continue

        print(f"Memuat Model Fold {fold}...")
        model = PatchDINOv2(num_classes=len(CFG.classes)).to(CFG.device)
        model.load_state_dict(torch.load(model_path, map_location=CFG.device))

        if CFG.use_multi_gpu:
            model = nn.DataParallel(model)
        model.eval()
        n_loaded += 1

        with torch.no_grad():
            for patches, img_ids in tqdm(test_loader, desc=f"Inferensi Fold {fold}"):
                patches = patches.to(CFG.device)
                with autocast():
                    logits, _ = model(patches)
                probs = F.softmax(logits, dim=1).cpu().numpy()
                for i, img_id in enumerate(img_ids):
                    ensemble_probs[img_id] += probs[i]

        del model
        torch.cuda.empty_cache()

    # Rata-ratakan akumulasi probabilitas
    print(f"\nTotal model dimuat: {n_loaded}/{CFG.n_fold}")
    print("=== Membuat File Submission ===")

    records = []
    for img_id, probs in ensemble_probs.items():
        pred_idx   = np.argmax(probs / max(n_loaded, 1))
        pred_label = CFG.idx_to_class[pred_idx]
        records.append({'id': img_id, 'label': pred_label})

    sub_df = (
        pd.DataFrame(records)
        .sort_values('id')
        .reset_index(drop=True)
    )
    sub_df.to_csv("submission.csv", index=False)
    print("✅ 'submission.csv' berhasil dibuat!")
    print(sub_df['label'].value_counts())
    display(sub_df.head(10))


run_inference_and_submit()
```
